## Usage Instructions

### Workflow

1. **Tab 1: System** - Create new system or load existing
   - Enter system name and select category
   - Add initial notes (optional)
   - Click **💾 Save Current** to persist data

2. **Tab 2: Architecture** - Assess AI/ML design (Criterion 1)
   - Select framework (Transformer, Diffusion, etc.)
   - Choose generation mode
   - Select modalities (hold Ctrl/Cmd for multiple)
   - Rate confidence (1-5)
   - Document evidence sources

3. **Tab 3: Interface** - Evaluate usability (Criterion 2)
   - Select interaction modes
   - Choose setup complexity
   - Estimate setup time
   - Rate confidence (1-5)
   - Document evidence sources

4. **Tab 4: Hardware** - Assess requirements (Criterion 3)
   - Select GPU type
   - Enter VRAM/RAM requirements
   - Assess accessibility levels
   - Rate confidence (1-5)
   - Document evidence sources

5. **Tab 5: Batch Ops** - Export and manage systems
   - View all assessed systems
   - Export to JSON or CSV
   - Import previous assessments
   - Generate completion report

### Data Storage

- **Individual Systems:** `outputs/phase1_results/{system_name}_phase0.json`
- **Consolidated Export:** `outputs/phase1_results/phase0_all_systems.json`
- **CSV Summary:** `outputs/phase1_results/phase0_summary.csv`
- **Completion Report:** `outputs/phase1_results/phase0_completion_report.txt`

### Tips

- **Confidence Ratings:** 1=Uncertain, 2=Low, 3=Moderate, 4=High, 5=Very High
- **Evidence:** Include links to documentation, papers, blog posts, or video tutorials
- **Save Often:** Click **💾 Save Current** after completing each criterion
- **Load Existing:** Use the dropdown in Tab 1 to continue previous assessments
