```python
# Cell 6: Store all widget references for data collection
logger = logging.getLogger('notebook')

logger.info("📚 Organizing widget references...")

# Create comprehensive widget reference dictionary
all_widgets = {
    'core_session': core_session['widgets'],
    'critical_incident': critical_incident['widgets'],
    'workflow_phase': workflow_phase['widgets']
}

# Display structure
logger.info("✅ Widget references organized:")
logger.debug(f"   Core Session sections: {list(all_widgets['core_session'].keys())}")
logger.debug(f"   Incident sections: {list(all_widgets['critical_incident'].keys())}")
logger.debug(f"   Workflow Phase sections: {list(all_widgets['workflow_phase'].keys())}")

# Count total widgets
def count_widgets(w_dict, depth=0):
    count = 0
    for key, value in w_dict.items():
        if key == 'container':
            continue
        elif isinstance(value, dict):
            count += count_widgets(value, depth+1)
        elif hasattr(value, 'value'):
            count += 1
    return count

total_widgets = count_widgets(all_widgets)
logger.info(f"📊 Total widgets available: ~{total_widgets}")

# ============================================================================
# SESSION STATE MANAGEMENT SETUP (v2.0.0)
# ============================================================================

logger.info("🔧 Setting up session state management...")

if shared_utils_available:
    # v2.0.0: Use new shared utilities module
    try:
        setup_session_management_observers(all_widgets, main_tabs)
        logger.info("✅ Session state management configured (v2.0.0 shared utilities)!")
    except Exception as e:
        logger.warning(f"Could not configure session management observers: {e}")
        logger.info("Session will still work, just without auto-save functionality")
else:
    # v1.0.0 fallback: Session management built into core_session_logging
    try:
        from core_session_logging import setup_session_management_observers
        setup_session_management_observers(all_widgets, main_tabs)
        logger.info("✅ Session state management configured (v1.0.0 fallback)!")
    except ImportError:
        logger.warning("Session management not available - basic mode")
        logger.info("Session will still work, just without auto-save functionality")

logger.info("💡 Use '🆕 New Session' to clear all data, '💾 Save State' to checkpoint progress")
```

```python
# Cell 7: Set up observers for dynamic behavior
logger = logging.getLogger('notebook')

logger.info("👁️ Setting up observers for dynamic content...")

try:
    # Observer for phase-specific content switching
    def update_phase_specific_content(change):
        """Update Phase-Specific Details tab content based on selected phase type."""
        phase_type = change['new']
        logger.debug(f"Phase type changed to: {phase_type}")
        
        # Get the phase-specific container and widgets
        phase_specific = all_widgets['workflow_phase']['phase_specific']
        container = phase_specific['phase_specific_container']
        
        if phase_type == 'Content Generation':
            gen_widgets = phase_specific['generation']
            container.children = [
                widgets.HTML('<h3 style="color: #2196F3; border-bottom: 2px solid #2196F3; padding-bottom: 10px;">⚡ Content Generation Details</h3>'),
                gen_widgets['gen_approach'],
                gen_widgets['gen_prompts'],
                gen_widgets['gen_iterations'],
                gen_widgets['gen_output']
            ]
            logger.debug("Content Generation tab content updated")
        elif phase_type == 'Curation':
            cur_widgets = phase_specific['curation']
            container.children = [
                widgets.HTML('<h3 style="color: #4CAF50; border-bottom: 2px solid #4CAF50; padding-bottom: 10px;">🎨 Curation Details</h3>'),
                cur_widgets['cur_candidates_reviewed'],
                cur_widgets['cur_candidates_selected'],
                cur_widgets['cur_criteria'],
                cur_widgets['cur_notes']
            ]
            logger.debug("Curation tab content updated")
        elif phase_type == 'Integration':
            int_widgets = phase_specific['integration']
            container.children = [
                widgets.HTML('<h3 style="color: #FF9800; border-bottom: 2px solid #FF9800; padding-bottom: 10px;">🔧 Integration Details</h3>'),
                int_widgets['int_activities'],
                int_widgets['int_challenges'],
                int_widgets['int_success']
            ]
            logger.debug("Integration tab content updated")
        elif phase_type == 'Post-Production':
            pp_widgets = phase_specific['post_production']
            container.children = [
                widgets.HTML('<h3 style="color: #FF9800; border-bottom: 2px solid #FF9800; padding-bottom: 10px;">🎚️ Post-Production Details</h3>'),
                pp_widgets['pp_activities'],
                pp_widgets['pp_ai_percentage'],
                pp_widgets['pp_modifications'],
                pp_widgets['pp_quality']
            ]
            logger.debug("Post-Production tab content updated")

    # Attach observer to phase_type widget
    phase_type_widget = all_widgets['workflow_phase']['setup']['phase_type']
    phase_type_widget.observe(update_phase_specific_content, names='value')

    # Initialize with current value
    update_phase_specific_content({'new': phase_type_widget.value})

    logger.info("✅ Observers configured successfully (v2.0.0):")
    logger.debug("   - Phase-specific content switching (Workflow Phase tab)")
    logger.info("💡 Workflow phase content will update automatically based on Phase Type selection")

except Exception as e:
    logger.error(f"Error setting up observers: {e}")
    logger.warning("Observers may not work correctly - continuing anyway")
```
