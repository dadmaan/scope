```python
# Cell 1: Imports
import ipywidgets as widgets
from IPython.display import display, HTML, Markdown, clear_output
from datetime import datetime, date
import json
from pathlib import Path
import logging

# Import widget modules from v2.0.0 package via compatibility layer
try:
    from compat_imports import (
        create_system_metadata_section,
        create_session_import_section,
        create_assessment_tabs,
        create_journal_tabs,
        create_validation_export_section,
        export_to_markdown,
        export_to_csv
    )
    print("✅ All imports successful from aimusic_eval package!")
    print("📦 Framework utilities loaded")
except ImportError as e:
    print(f"❌ Import error: {e}")
    print("Please ensure the aimusic_eval package is installed: pip install -e .")
    raise

# Create logs directory
logs_dir = Path.cwd().parent / 'outputs' / 'logs'
logs_dir.mkdir(parents=True, exist_ok=True)

# Configure root logger
logging.basicConfig(
    level=logging.INFO,
    format='%(asctime)s - %(name)s - %(levelname)s - %(message)s',
    datefmt='%Y-%m-%d %H:%M:%S'
)
root_logger = logging.getLogger()

# Create file handler
log_file = logs_dir / f"synthesis_{datetime.now().strftime('%Y%m%d_%H%M%S')}.log"
file_handler = logging.FileHandler(log_file)
file_handler.setLevel(logging.DEBUG)
file_formatter = logging.Formatter(
    '%(asctime)s - %(name)s - %(levelname)s - %(message)s',
    datefmt='%Y-%m-%d %H:%M:%S'
)
file_handler.setFormatter(file_formatter)
root_logger.addHandler(file_handler)

# Create notebook logger
notebook_logger = logging.getLogger('notebook')
notebook_logger.info(f"📝 Synthesis log file: {log_file}")

print("🎯 Synthesis Journal v2.0 Ready")
```

```python
# Cell 2: Global Data Structures (Notebook Scope - PRESERVED)

# Main synthesis data structure
synthesis_data = {
    'template_version': '2.0',
    'template_type': 'synthesis_journal',
    'created_at': datetime.now().isoformat(),
    'modified_at': datetime.now().isoformat(),
    'system_metadata': {},
    'imported_sessions': [],
    'quantitative_assessment': {
        'criteria': {},
        'average_score': 0,
        'synthesis': {}
    },
    'reflective_journal': {
        'entries': [],
        'entry_count': 0
    }
}

# Global state variables (notebook scope)
loaded_sessions = []  # Stores parsed session JSON data
criterion_widgets = {}  # Stores all criterion widget references

print("✅ Global data structures initialized")
print(f"📊 Synthesis data schema: {list(synthesis_data.keys())}")
```

```python
# Cell 3: Create All Widget Sections

print("🔨 Creating widget sections...")

# Create system metadata section
system_metadata = create_system_metadata_section()
print("✅ System metadata widgets created")

# Create session import section
session_import = create_session_import_section(loaded_sessions)
print("✅ Session import widgets created")

# Create quantitative assessment tabs (4 nested sub-tabs)
assessment_tabs = create_assessment_tabs(loaded_sessions, criterion_widgets)
print("✅ Quantitative assessment tabs created (4 sub-tabs)")

# Create reflective journal tabs (7 nested sub-tabs)
journal_tabs = create_journal_tabs(synthesis_data)
print("✅ Reflective journal tabs created (7 sub-tabs)")

print("\n🎉 All widget sections ready!")
print("📊 Total nested tabs: 11 (4 assessment + 7 journal)")
```
